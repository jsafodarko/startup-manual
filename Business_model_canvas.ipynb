```python
import ipywidgets as widgets
from IPython.display import display, HTML
import pandas as pd
from datetime import datetime

class BusinessModelCanvas:
    def __init__(self):
        # Create text areas with prompts and better sizing
        self.key_partners = widgets.Textarea(
            description='Key Partners:',
            placeholder='• Who are your key suppliers?\n• Which key resources do you acquire from partners?\n• Which key activities do partners perform?\n• What strategic alliances do you need?',
            layout=widgets.Layout(width='90%', height='200px')
        )

        self.key_activities = widgets.Textarea(
            description='Key Activities:',
            placeholder='• What key activities does your value proposition require?\n• What activities are needed for distribution channels?\n• What activities are needed for customer relationships?\n• What activities drive revenue?',
            layout=widgets.Layout(width='90%', height='150px')
        )

        self.key_resources = widgets.Textarea(
            description='Key Resources:',
            placeholder='• What key resources does your value proposition require?\n• What physical assets are needed?\n• What intellectual property is essential?\n• What human resources do you need?',
            layout=widgets.Layout(width='90%', height='150px')
        )

        self.value_propositions = widgets.Textarea(
            description='Value Props:',
            placeholder='• What value do you deliver to customers?\n• Which problems do you solve?\n• What bundles of products/services do you offer?\n• How do you differ from competition?',
            layout=widgets.Layout(width='90%', height='200px')
        )

        self.customer_relationships = widgets.Textarea(
            description='Customer Relationships:',
            placeholder='• What type of relationship does each customer segment expect?\n• How costly are they?\n• How are they integrated with your business model?',
            layout=widgets.Layout(width='90%', height='150px')
        )

        self.channels = widgets.Textarea(
            description='Channels:',
            placeholder='• Through which channels do customers want to be reached?\n• How are you reaching them now?\n• Which channels work best?\n• Which are most cost-efficient?',
            layout=widgets.Layout(width='90%', height='150px')
        )

        self.customer_segments = widgets.Textarea(
            description='Customer Segments:',
            placeholder='• For whom are you creating value?\n• Who are your most important customers?\n• What are their characteristics?\n• Where are they located?',
            layout=widgets.Layout(width='90%', height='200px')
        )

        self.cost_structure = widgets.Textarea(
            description='Cost Structure:',
            placeholder='• What are the most important costs?\n• Which key resources are most expensive?\n• Which key activities are most expensive?\n• Fixed vs variable costs?',
            layout=widgets.Layout(width='90%', height='150px')
        )

        self.revenue_streams = widgets.Textarea(
            description='Revenue:',
            placeholder='• For what value are customers willing to pay?\n• How do they currently pay?\n• How would they prefer to pay?\n• How much does each revenue stream contribute?',
            layout=widgets.Layout(width='90%', height='150px')
        )
        # Create save and export buttons
        self.save_button = widgets.Button(
            description='💾 Save to Excel',
            layout=widgets.Layout(width='150px')
        )
        self.save_button.on_click(self.save_to_excel)

        self.clear_button = widgets.Button(
            description='🗑️ Clear All',
            layout=widgets.Layout(width='150px')
        )
        self.clear_button.on_click(self.clear_all)

    def display_canvas(self):
        # Display title
        display(HTML('<h1>Business Model Canvas</h1>'))
        display(HTML('<p style="color: #666;">Click inside each box to see guiding questions</p>'))

        # Create layout
        top_row = widgets.HBox([self.key_partners, self.value_propositions, self.customer_segments])
        middle_left = widgets.VBox([self.key_activities, self.key_resources])
        middle_right = widgets.VBox([self.customer_relationships, self.channels])
        middle_section = widgets.HBox([middle_left, middle_right])
        bottom_row = widgets.HBox([self.cost_structure, self.revenue_streams])
        buttons_row = widgets.HBox([self.save_button, self.clear_button])

        # Display all elements
        display(top_row)
        display(middle_section)
        display(bottom_row)
        display(buttons_row)

    def save_to_excel(self, b):
        # Create a dictionary of all sections
        data = {
            'Section': ['Key Partners', 'Key Activities', 'Key Resources',
                       'Value Propositions', 'Customer Relationships', 'Channels',
                       'Customer Segments', 'Cost Structure', 'Revenue Streams'],
            'Content': [self.key_partners.value, self.key_activities.value,
                       self.key_resources.value, self.value_propositions.value,
                       self.customer_relationships.value, self.channels.value,
                       self.customer_segments.value, self.cost_structure.value,
                       self.revenue_streams.value]
        }

        # Convert to DataFrame
        df = pd.DataFrame(data)

        # Generate filename with timestamp
        timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
        filename = f'business_model_canvas_{timestamp}.xlsx'

        # Save to Excel
        df.to_excel(filename, index=False)
        print(f'✅ Saved to {filename}')

    def clear_all(self, b):
        # Clear all text areas
        self.key_partners.value = ''
        self.key_activities.value = ''
        self.key_resources.value = ''
        self.value_propositions.value = ''
        self.customer_relationships.value = ''
        self.channels.value = ''
        self.customer_segments.value = ''
        self.cost_structure.value = ''
        self.revenue_streams.value = ''
        print('🔄 All fields cleared')
```

```python
canvas = BusinessModelCanvas()
canvas.display_canvas()
```
